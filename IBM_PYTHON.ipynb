**CLASS AND OBJECT**

**Class in Python**
A **class** is a blueprint or template used to create objects.
It groups data (variables) and behaviors (functions) into a single unit.

### **Syntax:**

```
class ClassName:
    # variables
    # methods
```

**Object in Python**
An **object** is an instance of a class.
It represents a real thing created from a class and has its own data and behavior.

### **Syntax:**

```
object_name = ClassName()
```

**self in Python**
`self` refers to the **current object** inside a class.
It is used to access object variables and methods.
Python automatically passes `self` as the first argument in methods.

### **Syntax:**

```
def methodName(self):
    # statements
```

**Inheritance in Python**
Inheritance allows one class (child) to acquire the properties and methods of another class (parent).
It supports code reuse and extensibility.

### **Syntax:**

```
class ChildClass(ParentClass):
    # new properties or methods
```

````python
#WAP TO CREATE CLASS AND OBJECT

class A:
  def __init__(self,name,age):
    self.name = name
    self.age = age
  def display(self):
    print("Name=", self.name)
    print("Age=", self.age)

A1 = A("John",25)
A1.display()
````

````python
#WAP to find area of rectangle

class rectangle:
  def __init__(self,length,width):
    self.length = length
    self.width = width
  def area(self):
    return self.length*self.width

rectangle = rectangle(5,4)
print(rectangle.area())
````

**1. Multiple Inheritance in Python**

Multiple inheritance means a class can inherit from **more than one parent class**.
The child class gets the features (methods and variables) of **all** parent classes.

### **Syntax:**

```
class ChildClass(Parent1, Parent2, Parent3):
    # body of child class
```

**2. Multilevel Inheritance in Python**

Multilevel inheritance means inheritance happens in a **chain**, like
**Grandparent → Parent → Child**.
A class inherits from another class, which itself inherits from another class.

### **Syntax:**

```
class A:
    # base class

class B(A):
    # B inherits A

class C(B):
    # C inherits B (multilevel)
```


````python
#WAP to demostrate the multiple inheritance

class camera:
  def __init__(self,camera):
    self.camera = camera

class speaker:
  def __init__(self,speaker):
    self.speaker = speaker

class battery:
  def __init__(self,battery):
    self.battery = battery

class phone(camera,speaker,battery):
  def __init__(self,camera,speaker,battery):
    self.camera = camera
    self.speaker = speaker
    self.battery = battery

xiaomi = phone("100 MP", "Dolby Atmos", "7000 mah")
print("features of xiaomi phone are:")
print("Camera: ", xiaomi.camera)
print("Speaker:" , xiaomi.speaker)
print("Battery:" , xiaomi.battery)
````

# **Pandas**

**Pandas** is one of the most popular and powerful **Python libraries used for data analysis and data manipulation**.

Pandas helps you work with data easily — especially **tables, spreadsheets, CSV files, databases**, and any kind of structured data.

It comes from the term **“Panel Data”**, which refers to multi-dimensional structured datasets used in statistics/econometrics.

**What Pandas Provides**
Pandas mainly offers two powerful data structures:

1. **Series**
* A one-dimensional data structure (like a column in Excel).
* Example: `['A', 'B', 'C']`

2. **DataFrame**
* A two-dimensional data structure (like an Excel sheet with rows & columns).
* Example:

| Name | Age |
| ---- | --- |
| Alex | 21  |
| Bob  | 25  |

**What Can You Do With Pandas?**

Pandas makes data handling extremely easy:

* Read data from CSV, Excel, SQL, JSON
* Filter and clean data
* Handle missing values
* Sort and reshape data
* Perform group-by and aggregations
* Merge and join datasets
* Analyze large datasets efficiently

**MERGING, JOINING AND CONCATENATION**\

**1. CONCATENATION**

**Meaning:**
Putting two datasets **together**, either **one below the other** or **side by side**.

When you use it:
* When two tables have **same columns** and you want **more rows**.
* When two tables have **same rows** and you want **more columns**.

Example (Row-wise):
Table 1 has students A, B
Table 2 has students C, D
→ Concatenation gives A, B, C, D.

Example (Column-wise):
Left table has names
Right table has marks
→ Concatenation puts marks next to names.

````python
import pandas as pd
````

````python
df1=pd.DataFrame({'A':['A0','A1','A2','A3'],
                  'B':['B0','B1','B2','B3'],
                  'C':['C0','C1','C2','C3']},
                 index=[0,1,2,3])
````

````python
df2=pd.DataFrame({'A':['A4','A5','A6','A7'],
                  'B':['B4','B5','B6','B7'],
                  'C':['C4','C5','C6','C7']},
                  index=[4,5,6,7])
````

````python
df3=pd.DataFrame({'A':['A8','A9','A10','A11'],
                  'B':['B8','B9','B10','B11'],
                  'C':['C8','C9','C10','C11']},
                  index=[8,9,10,11])
````

````python
pd.concat([df1,df2,df3],axis=1)
````

````python
pd.concat([df1,df2,df3],axis=0)
````

**2. MERGING**

**Meaning:**
Combining two tables using a **common column** (like SQL joins).

Common column example:
Both tables have a column “id”.

Types of Merges:
1. **Inner Merge** – Only the matching rows are included.
2. **Left Merge** – All rows from left table, matched or not.
3. **Right Merge** – All rows from right table.
4. **Outer Merge** – All rows from both tables.

Real-life example:

Table 1: Student Name + ID
Table 2: Student ID + Marks

→ Merge using ID.
→ It connects names with marks.

join_diagram.93e6ef63afbe.png

````python
pd.merge(left,right,how='inner',on='key')
````

````python
pd.merge(left,right,how='left',on='key')
````

````python
pd.merge(left,right,how='right',on='key')
````

````python
pd.merge(left,right,how='outer',on='key')
````

**3. JOINING**

**Meaning:**

Joining two tables using their **index** (row labels), not columns.

When you use it:
* When both tables are arranged by index (like names as index).
* You want to attach extra information based on index.

Example:
Table 1:
Ajay → Age
Riya → Age

Table 2:
Ajay → Marks
Riya → Marks

Joining matches the rows because both have the same index (Ajay, Riya).

````python
left = pd.DataFrame({"A": ["A0", "A1", "A2"], "B": ["B0", "B1", "B2"]}, index=["K0", "K1", "K2"])
right = pd.DataFrame({"C": ["C0", "C2", "C3"], "D": ["D0", "D2", "D3"]}, index=["K0", "K2", "K3"])
````

# **DIFFERENCE BETWEEN SHALLOW COPY AND DEEP COPY**

| Feature                    | **Shallow Copy**                                                      | **Deep Copy**                                                       |
| -------------------------- | --------------------------------------------------------------------- | ------------------------------------------------------------------- |
| **Definition**             | Creates a new object but **copies references** of nested objects.     | Creates a new object and **recursively copies all nested objects**. |
| **Nested Objects**         | **Shared** between original and copy.                                 | **Independent** from the original.                                  |
| **Changes in Nested Data** | Changes affect both original and copy.                                | Changes do **not** affect each other.                               |
| **Performance**            | Faster, uses less memory.                                             | Slower, uses more memory (because everything is copied).            |
| **Use When**               | When object is simple OR nested elements don’t need full duplication. | When you need a completely independent clone.                       |


````python
#shallow copy example
import copy
list1 = [[1, 2], [3, 4]]
list2 = copy.copy(list1)

list2[0][0] = 100
print(list1)  # [[100, 2], [3, 4]]
````

````python
#deep copy example
import copy
list1 = [[1, 2], [3, 4]]
list2 = copy.deepcopy(list1)

list2[0][0] = 100
print(list1)  # [[1, 2], [3, 4]]
````
